```python
import cv2
import numpy as np
```

```python
# Video settings
width, height = 640, 480
fps = 20
```

```python
video = cv2.VideoWriter(
    "ai_animation.mp4",
    cv2.VideoWriter_fourcc(*'mp4v'),
    fps,
    (width, height)
)
```

```python
# Generate animation frames
for i in range(120):
    frame = np.zeros((height, width, 3), dtype=np.uint8)

    # Moving circle animation
    x_position = 50 + i * 4

    cv2.circle(
        frame,
        (x_position, height // 2),
        40,
        (0, 255, 255),
        -1
    )

    # Add AI-style label
    cv2.putText(
        frame,
        "AI Generated Animation",
        (150, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255,255,255),
        2
    )

    video.write(frame)
```

```python
video.release()

print("Video animation generated successfully!")
```
