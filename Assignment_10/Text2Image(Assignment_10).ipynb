```python
import torch
from diffusers import StableDiffusionPipeline
```

```python
# Select device
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5"
)
pipe = pipe.to(device)
```

```python
# Text prompt
prompt = "A professional close-up of a mahogany desk with a silver 'LLM' nameplate, a classic leather-bound law book, and a judge's gavel, soft sunlight through a library window."
```

```python
# Generate image
image = pipe(prompt).images[0]
```

```python
# Save output
image.save("generated_image.png")

print("Image generated successfully!")
```
